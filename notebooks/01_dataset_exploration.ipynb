```python
from datasets import load_dataset

dataset = load_dataset(
    "Project-AgML/tropical_flower_variety_classification",
    split="train",
    streaming=True
)

print(dataset.features)
```

```python
from collections import Counter

counts = Counter()

for example in dataset:
    counts[example["label"]] += 1

print(dict(counts))
```

```python
import requests

url = (
    "https://datasets-server.huggingface.co/first-rows"
    "?dataset=Project-AgML%2Ftropical_flower_variety_classification"
    "&config=default&split=train"
)

response = requests.get(url, timeout=30)

print("Status:", response.status_code)

if response.ok:
    result = response.json()
    print("Columns:", [column["name"] for column in result["features"]])
else:
    print(response.text[:500])
```

```python
import requests

url = (
    "https://datasets-server.huggingface.co/rows"
    "?dataset=Project-AgML%2Ftropical_flower_variety_classification"
    "&config=default&split=train&offset=0&length=5"
)

response = requests.get(url, timeout=30)
print("Status:", response.status_code)

if response.ok:
    rows = response.json()["rows"]
    for item in rows:
        print(item["row"]["label"])
else:
    print(response.text[:500])
```

```python
import requests

url = "https://datasets-server.huggingface.co/filter"
params = {
    "dataset": "Project-AgML/tropical_flower_variety_classification",
    "config": "default",
    "split": "train",
    "where": '("label" = 2)',
    "length": 5,
}

try:
    response = requests.get(url, params=params, timeout=90)
    print("Status:", response.status_code)
    print(response.text[:1000])
except requests.exceptions.RequestException as e:
    print("Request failed:", e)
```

```python
import requests

url = (
    "https://datasets-server.huggingface.co/rows"
    "?dataset=Project-AgML%2Ftropical_flower_variety_classification"
    "&config=default&split=train&offset=0&length=5"
)

try:
    response = requests.get(url, timeout=90)
    print("Status:", response.status_code)
    if response.ok:
        print("Rows returned:", len(response.json()["rows"]))
    else:
        print(response.text[:300])
except requests.exceptions.RequestException as e:
    print(type(e).__name__, e)
```

```python
import requests

url = "https://datasets-server.huggingface.co/filter"
params = {
    "dataset": "Project-AgML/tropical_flower_variety_classification",
    "config": "default",
    "split": "train",
    "where": 'label="2"',
    "length": 5,
}

try:
    response = requests.get(url, params=params, timeout=90)
    print("Status:", response.status_code)
    print(response.text[:800])
except requests.exceptions.RequestException as e:
    print("Request failed:", e)
```
